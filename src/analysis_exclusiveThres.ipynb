```python
# view all traces at 80 dB pre-ZX1

import lib.signalProcess as signalProcess
import os

# import or use newly-generated ROI
roi_mask = dmask['mask']
# roi_mask = joblib.load(os.path.join(expDir, 'response_mask.joblib'))

t = signalProcess.getTimeVec(rawF.shape[-1])
F_linFilt, slope, intercept = signalProcess.subtractLinFit(t, rawF[:, roi_mask==1, :].mean(axis=(0,1)), t_base = t_base)

fig,ax = plt.subplots(3, 1, figsize=(15,10))

ax[0].plot(t, rawF[:, roi_mask==1, :].mean(axis=1).T)
ax[0].set_ylabel('raw F')
ax[0].set_title("raw F individual traces")

ax[1].plot(t, rawF[:, roi_mask==1, :].mean(axis=(0,1)), 'k-')
ax[1].plot(t, t*slope+intercept, 'b-', label='linear fit')
ax[1].set_title('raw F trace mean')
ax[1].set_ylabel('raw F')

ax[2].plot(t, F_linFilt, 'k-')
ax[2].set_title('raw F trace mean after linear fit subtraction')
ax[2].set_ylabel('raw F - linear fit + baseline')
ax[2].set_xlabel('time (s)')

for axn in ax:
    # set stimulus start point
    axn.axvline(x=t_base[1], color='r', linestyle=':')
    # get the y-axis limits
    ymin, ymax = axn.get_ylim()
    # show baseline and response time window
    axn.fill_betweenx((ymin, ymax), t_base[0], t_base[1], color='g', alpha=0.2, label='baseline window')
    axn.fill_betweenx((ymin, ymax), t_resp[0], t_resp[1], color='r', alpha=0.2, label='response window')

ax[1].legend()

fig.suptitle(f"{expDir}: {dB} dB pre-ZX1")
plt.show()
```

```python
# add ROI mask, time vector, raw F, f0, dF, and dFF for each trace into metadata dataframe
# only consider fluorescence response within ROI
# negative outliers marked by `None` in columns `dF_ROI_linFilt_negExcl_butterFilt_peak`, and `dFF_ROI_linFilt_negExcl_butterFilt_peak`

# set time window to apply negative exclusive thresholds
# suggested overlapping with response time window
t_resp_excl=t_resp

# set cutoff frequency for low-pass filter
cutoff_freq = 2


# # add binary mask of ROI
df_drop['ROImask'] = df_drop.apply(lambda x: roi_mask, axis=1)

# # add time vector
df_drop['time'] = df_drop.apply(lambda x: t, axis=1)

# # add fluorescence trace (F) within ROI
# raw data
df_drop['F_ROI_raw'] = df_drop.apply(lambda x: qcam2img[x['qcam']][roi_mask==1, :].mean(axis=0), axis=1)
# processed data (subtracted linear fit and added low-pass filter)
df_drop['F_ROI_linFilt_butterFilt'] = df_drop['F_ROI_raw'].apply(lambda x: signalProcess.butterFilter(signalProcess.subtractLinFit(t, x, t_base=t_base)[0], cutoff_freq=cutoff_freq))

# # add baseline fluorescence (f0) within ROI
# raw data
df_drop['f0_ROI_raw'] = df_drop['F_ROI_raw'].apply(lambda x: signalProcess.getBaseResp(x, t, t_base=t_base)[0])
# processed data (subtracted linear fit and added low-pass filter)
df_drop['f0_ROI_linFilt_butterFilt'] = df_drop['F_ROI_linFilt_butterFilt'].apply(lambda x: signalProcess.getBaseResp(x, t, t_base=t_base)[0])

# # add dF response within ROI
# raw data
df_drop['dF_ROI_raw'] = df_drop.apply(lambda x: x['F_ROI_raw'] - x['f0_ROI_raw'], axis=1)
df_drop['dF_ROI_raw_peak'] = df_drop.apply(lambda x: signalProcess.pkDFFimg(qcam2img[x['qcam']], subLinFit=False, ROImask=roi_mask, butterFilt=False, dFResp=True, 
                                                                            negExcl=False, t_base=t_base, t_resp=t_resp, t_resp_excl=t_resp_excl), axis=1)
# processed data (subtracted linear fit, excluded negative traces, and added low-pass filter)
df_drop['dF_ROI_linFilt_butterFilt'] = df_drop.apply(lambda x: x['F_ROI_linFilt_butterFilt'] - x['f0_ROI_linFilt_butterFilt'], axis=1)
df_drop['dF_ROI_linFilt_negExcl_butterFilt_peak'] = df_drop.apply(lambda x: signalProcess.pkDFFimg(qcam2img[x['qcam']], ROImask=roi_mask, cutoff_freq=cutoff_freq, dFResp=True, 
                                                                                                   t_base=t_base, t_resp=t_resp, t_resp_excl=t_resp_excl), axis=1)

# # add dFF response within ROI
# raw data
df_drop['dFF_ROI_raw'] = df_drop.apply(lambda x: x['dF_ROI_raw'] / x['f0_ROI_raw'], axis=1)
df_drop['dFF_ROI_raw_peak'] = df_drop.apply(lambda x: signalProcess.pkDFFimg(qcam2img[x['qcam']], subLinFit=False, ROImask=roi_mask, butterFilt=False, 
                                                                             negExcl=False, t_base=t_base, t_resp=t_resp, t_resp_excl=t_resp_excl), axis=1)
# processed data (subtracted linear fit, excluded negative traces, and added low-pass filter)
df_drop['dFF_ROI_linFilt_butterFilt'] = df_drop.apply(lambda x: x['dF_ROI_linFilt_butterFilt'] / x['f0_ROI_linFilt_butterFilt'], axis=1)
df_drop['dFF_ROI_linFilt_negExcl_butterFilt_peak'] = df_drop.apply(lambda x: signalProcess.pkDFFimg(qcam2img[x['qcam']], ROImask=roi_mask, cutoff_freq=cutoff_freq, 
                                                                                                    t_base=t_base, t_resp=t_resp, t_resp_excl=t_resp_excl), axis=1)

# show truncated columns
pd.set_option('display.max_columns', 25)
pd.set_option('display.max_rows', 5)
df_drop
```
